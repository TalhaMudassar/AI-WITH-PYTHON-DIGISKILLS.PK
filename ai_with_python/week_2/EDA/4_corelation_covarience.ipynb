```python
# Load dataset
import pandas as pd

df = pd.read_csv("ecommerce_sales_analytics_5000.csv", parse_dates=["order_date"])
```

```python
numeric_df = df[["quantity", "unit_price", "discount", "delivery_days", "revenue"]]
# Covariance matrix
print("\nCovariance:\n", numeric_df.cov())
```

```python
# Correlation Matrix
print("\nCorrelation:\n", numeric_df.corr())
```

```python
numeric_df.corr()["revenue"].sort_values(ascending=False)         #Focus on Revenue Relationships
```

```python
numeric_df.corr().values.min(), numeric_df.corr().values.max()      #Verification
# Expected: Minimum >= -1 and Maximum <= +1
```
