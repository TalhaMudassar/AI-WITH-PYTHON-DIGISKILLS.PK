### Reading from & Writing to Files in Python

### Reading from file

```python
file = open("demo.txt", "r")
print(file.read())
file.close()
```

---

### Writing from file

```python
file = open("demo.txt", "w")
file.write("This is another test.")
file.close()
```

---

# Modes of File Handling - Read, Write & Append

Read Mode ("r")

```python
file1 = open("demo.txt", "r")
print(file1.read())
file1.close()
```

---

Write Mode ("w")

```python
file1 = open("demo.txt", "w")
file1.write("This is another test.")
file1.close()
```

---

Append Mode ("a")

```python
file1 = open("demo.txt", "a")
file1.write("\nNew line added!")
file1.close()
```
