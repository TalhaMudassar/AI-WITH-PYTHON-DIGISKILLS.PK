```python
# First Block
age = 17
country = "Pakistan"
if age >= 18:
    if country == "Pakistan":    # nested if
        print("You can vote in Pakistan.")
```

```python
if age > 18 and country == "Pakistan":    # combine conditions
    print("Eligible voter")
else:
    print("Not eligible voter")
```
