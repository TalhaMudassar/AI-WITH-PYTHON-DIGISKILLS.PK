# String METHODS

## .Lower()

```python
name = "Talha Mudassar"
result = name.lower()
print(f"Name are {result}")
```

```python
greeting = "HELLO Talha!"
print(greeting.lower()) 
```

## .UPPER()

```python
name = "Talha Mudassar"
result = name.upper()
print("Name converted into upper are ", result)
```

```python
warning = "do not enter"
print(warning.upper())  
```

# .STRIP()
### Removes extra whitespace (spaces, tabs, newlines) from the beginning and end of a string. Incredibly useful for cleaning up user input.

```python
name = "                   TALHA MUDASSAR"
print(f"Original Name are {name}")
#here use strip()
strip_name = name.strip()
print(f"Strip() name are : {strip_name}")
```

# .split()
### Breaks a string apart into a List of strings based on a specific separator (like a comma or a space).

```python
data = "banana,apple,strawberry"
data_list = data.split(",")
print(data_list)
```

# .join()
### The exact opposite of split(). It takes an iterable (like a List) and glues the items together into a single string, using a specific separator.

```python
naming_lists = ["hello","kesa","ho","saab"]
sentences = ",".join(naming_lists)
print(sentences)
```

```python
words = ["Python", "is", "awesome"]
sentence = " ".join(words) # Glues them together with a space
print(sentence) 
```

# . replace()
### Swaps out a specific phrase or character with a new one. Remember, this returns a new string; it doesn't change the original!

```python
sentence = "i hate you"
edit_sentence = sentence.replace("hate","love")
print(edit_sentence)
```

# find()
### Searches for a specific character or word and returns the index of its first occurrence. If it cannot find it, it safely returns -1.

```python
email = "talha@example.com"
at_symbol_index = email.find("@")
print(f"the @ symbol is at index {at_symbol_index}")
```

# . startswith()
### Returns True or False by checking if the string starts with a specific prefix.

```python
website = "https://google.com"
if website.startswith("http"):
    print("secure website")
else:
    print("unsecure website")
```

# .endswith()
### Returns True or False by checking if the string ends with a specific suffix. Great for checking file types!

```python
filename = "my_notes.pdf"
if filename.endswith(".pdf"):
    print("its .pdf file ")
else:
    print("its not a .pdf file")
```
