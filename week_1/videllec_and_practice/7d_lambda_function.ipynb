## map() and filter() are two very common Python functions used together with lambda. The easiest way to understand them is:

### map() → transform/change every item
### filter() → select/keep only items that satisfy a condition
### lambda → a small anonymous function that tells map() or filter() what to do

---
### map()

```python
numbers = [1,2,3,4,5]

result = map(lambda x: x*2,numbers)
print(list(result))
```

---
### filter()

It doesn't transform every item.
Instead, it checks every item and keeps only the items for which the condition is True.

```python
numbers = [1,2,3,4,5,6]

result = filter(lambda x: x%2 == 0, numbers)

print(list(result))
```
