### DateTime and Series 

```python
import pandas as pd
```

```python
df = pd.read_csv("ecommerce_sales_analytics_5000.csv",parse_dates=["order_date"])

# Setting datatime index
df = df.set_index("order_date")

df.index
```

```python
# Monthly Revnue Analysis
monthly_revnue = df["revenue"].resample("ME").sum()
monthly_revnue
```

```python
# Trend Interpretation
monthly_revnue.pct_change()
```

```python
monthly_revnue.index.is_monotonic_increasing  #Verify
```

```python
print(monthly_revnue.sum(), df["revenue"].sum())    #Check
```

```python
monthly_revnue.sum() <= df["revenue"].sum()     #Check
```
