```python
import pandas as pd
import openpyxl

df = pd.read_csv(
    'retail_sales.csv',
    parse_dates=['Date'],
    dtype={
        "Category":"category",
        "Region":"category"
    }
    )
df.info()
```

```python
clean_df = df[df['Sales'] > 0 ]  #Remove Invalid Rows
subset = clean_df[["Date","Category","Sales","Region"]]
```

```python
subset.to_excel("subset_sales.xlsx", sheet_name='sales')
```

```python
subset.to_json(
    "subset_sales.json",
    orient="records",
    date_format="iso"
)

print("Files Exported !")
subset.info()
```
