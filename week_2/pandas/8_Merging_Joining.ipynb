### Merging and Joining in pandas

```python
import pandas as pd
```

```python
customers = pd.DataFrame({
    'CustomerID': [1,2,3],
    'Name': ['Alice','Bob','Charlie']
})

orders = pd.DataFrame({
    'OrderID': [101,102,103,104],
    'CustomerID': [1,2,2,3],
    'Amount': [250, 150, 200, 300]
})
```

```python
# Inner merge on CustomerID
df_inner = pd.merge(customers,orders, on='CustomerID', how='inner')
print("\n Inner Merge: \n",df_inner)
```

```python
# Left Merge in CustomerID
df_left = pd.merge(customers,orders,on='CustomerID',how='left')
print("\n Left Merge: \n",df_left)
```

```python
# Concatenate row examples
df_concat = pd.concat([customers,customers])
print("\n Concatenated Rows:\n",df_concat)
```
