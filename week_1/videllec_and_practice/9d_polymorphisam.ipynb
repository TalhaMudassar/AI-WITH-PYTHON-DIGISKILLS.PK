```python
class Dog:
    def speak(self):
        print("Woof!")
```

```python
class Cat:
    def speak(self):
        print("Meow!")
```

```python
for pet in [Dog(), Cat()]:
    pet.speak()
```

### over riding

```python
class Vehicle:
    def move(self):
        print("Vehicle is moving")
```

```python
class Car(Vehicle):
    def move(self):
        print("Car is driving")
```

```python
car1 = Car()
```

```python
car1.move()
```

### over loading

```python
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c
```

```python
calc1 = Calculator()
```

```python
print(calc1.add(5))
```

```python
print(calc1.add(5, 10))
```

```python
print(calc1.add(5, 10, 15))
```
