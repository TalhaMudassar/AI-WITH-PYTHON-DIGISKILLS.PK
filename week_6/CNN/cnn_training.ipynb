```python
import tensorflow as tf
from tensorflow.keras import layers, models
```

```python
mnist = tf.keras.datasets.mnist
(X_train,X_test), (y_train,y_test) = mnist.load_data()

# Normalize the pixel values
X_train,X_test = X_train / 255.0,   X_test / 255.0

# Add channel  dimesion
X_train = X_train[..., tf.newaxis]
X_test = X_test[..., tf.newaxis]
```

```python
model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.summary()
```

```python
model.compile(optimizer='adam',
              loss = 'sparse_categorical_crossentropy',
              metrics =  ['accuracy'])
```

```python
(_, original_y_train), (_, _) = tf.keras.datasets.mnist.load_data()
history = model.fit(X_train, original_y_train, epochs=5, validation_split=0.1)
```

```python
# Reload the MNIST dataset to get the correct test set
(_, _), (correct_X_test_raw, correct_y_test) = tf.keras.datasets.mnist.load_data()

# Apply the same normalization and channel dimension addition as done previously for features
correct_X_test = correct_X_test_raw / 255.0
correct_X_test = correct_X_test[..., tf.newaxis]

test_loss, test_acc = model.evaluate(correct_X_test, correct_y_test)
print(f"Test accuracy: {test_acc:.4f}%")
```
