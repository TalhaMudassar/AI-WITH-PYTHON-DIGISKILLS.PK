```python
#Load the Healthcare Dataset
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("healthcare_patient_analytics_seaborn.csv")
```

```python
df = df.sample(500)
#Axes-level Example (scatterplot)
sns.scatterplot(
    data=df,
    x="treatment_cost",
    y="recovery_score",
    hue="visit_type"
)
plt.title("Treatment Cost vs Recovery Score")
plt.show()
```

```python
#Apply Seaborn Theme
sns.set_theme(style="whitegrid")          #Always set the theme before plotting.
```

```python
#Recreating a Matplotlib Plot in Seaborn
```

```python
#Figure-level Example (relplot)
sns.relplot(
    data=df,
    x="treatment_cost",
    y="recovery_score",
    hue="visit_type",
    col="gender"
)
```

```python
#Apply Seaborn Theme
sns.set_theme(style="whitegrid")          #Always set the theme before plotting.
```

```python
#Recreating a Matplotlib Plot in Seaborn

#Matplotlib
plt.scatter(df["treatment_cost"], df["recovery_score"])
plt.xlabel("Treatment Cost")
plt.ylabel("Recovery Score")
```
