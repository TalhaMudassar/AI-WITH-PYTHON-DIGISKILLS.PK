# FOR LOOPS

```python
for i in range(5):
    print(i)
```

```python
for char in "python":
    print(char)
```

```python
i = 4
rows = i
cols = i

matrix = []
for r in range(rows):
    row = []
    for c in range(cols):
        row.append(0)
    matrix.append(row)

for row in matrix:
    print(' '.join(map(str, row)))
```

```python
fruits = ["apple", "banana","cheery"]
for x in fruits:
    print(x)
```

```python
for x in "banana":
    print(x)
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
  if x == "banana":
    break
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    continue
  print(x)
```

```python
for x in range(6):
  print(x)
```

```python
for x in range(2, 6):
  print(x)
```

```python
for x in range(2, 30, 3):
  print(x)
```

```python
for x in range(6):
  print(x)
else:
  print("Finally finished!")
```

```python
for x in range(6):
  if x == 3: break
  print(x)
else:
  print("Finally finished!")
```

```python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
  for y in fruits:
    print(x, y)
```

### for loop simple

```python
fruits = ["apple","banana","cheery"]
for f in fruits:
    print(f)
```

### Looping Through a String

```python
for x in "banana":
    print(x)
```

### The break Statement

```python
fruits = ["apple","banana","cheery"]
for f in fruits:
    if f == "banana":
        break
    print(f)
    
```

### The continue Statement

```python
fruits = ["apple","banana","cheery"]
for f in fruits:
    if f == "banana":
        continue
    print(f)
```

### The range() Function

```python
for x in range(10):
    print(x)
```

```python
for x in range(1,11,1):
    print(x)
```

```python
for y in range(1,20,2):
    print(y)
    
```

### for loop with else

```python
for x in range(20):
    print(x)
else:
    print("loop finished")
```

### Nested Loops

```python
first = ["big","premium","tasty"]
fruits = ["apple","cheery","strawberry"]

for f in first:
    for fr in fruits:
        print(f,fr)
```

### The pass Statement

```python
for x in range(10):
    pass
```
