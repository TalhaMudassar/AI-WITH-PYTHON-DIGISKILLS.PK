### Custome Exceptions

```python
class InvalidAgeError(Exception):
    pass
```

```python
age = int(input("Enter the age"))

if age < 0:
    raise InvalidAgeError(f"Age cannot be negtive! , {age}")
```

```python
try:
    age = int(input("Enter the age"))
    
    if age < 0:
        raise InvalidAgeError(f"Age cannot be negtive,{age}")
except InvalidAgeError as e:
    print(e)
    
```
