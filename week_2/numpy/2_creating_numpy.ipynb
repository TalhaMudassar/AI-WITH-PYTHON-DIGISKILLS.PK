## Create the Numpy Array

```python
import numpy as np
```

```python
arr = np.array([1,2,3])
print(arr)
```

```python
print(np.zeros((2,3)))   # all row and col in array are zero
```

```python
print(np.ones((4,2))) # all the row and col in array are one we use this in multiplication
```

```python
print(np.full((2,3),7))
```

```python
print(np.arange(0,10,2))
```

```python
print(np.linspace(0,1,5))
```

```python
print(np.eye(3))
```

```python
print(np.diag([1,2,3]))
```

```python
np.array([1,2,3],dtype=float)
```

```python
arr = np.array([1.5,2.7,3.9])
print(arr.astype(int))
```
