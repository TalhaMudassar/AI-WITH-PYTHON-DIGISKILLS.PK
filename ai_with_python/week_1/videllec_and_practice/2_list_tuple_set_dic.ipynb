# list

```python
fruits = ["apple","banana","mango","banana"]
print(fruits)
```

### Accessing element

```python
print(fruits[0])
print(fruits[-1])
```

### List Methods (Most Important)

```python
fruits.append("orange")       # Add item
print(fruits)

fruits.insert(1, "grapes")    # Insert at index
print(fruits)

fruits.remove("banana")       # Remove item
print(fruits)

fruits.pop()                  # Remove last item
print(fruits)

fruits.sort()                 # Sort list
print(fruits)


```

### Looping Through List


```python
for fruit in fruits:
    print(fruit)
```

# tuples

### What is a Tuple?
A tuple is an ordered but unchangeable collection.

✔ Ordered  ✖ Cannot modify after creation

```python
colors = ("red", "green", "blue")
print(colors)
```

### Access Tuple

```python
print(colors[1])
```

### Tuple Properties

```python
# Tuples are immutable (cannot be changed)
colors[0] = "black"  # This will cause an error
```

### Tuple Methods

```python
numbers = (1,2,3,4,5,6,7,4,4)

print(numbers.index(5))
print(numbers.count(4))
```

# set

### What is a Set?
A set stores unique unordered values.

✔ No duplicates ✔ Fast operations

```python
my_set = {1, 2, 3, 4, 4, 5}
print(my_set)    # No duplicates
```

### Set Methods

```python
my_set.add(6)
my_set.remove(2)
print(my_set)
```

### Set Operations

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
```

# dictionary

### What is a Dictionary?
Stores data as key : value pairs.

```python
student = {
    "name": "Ali",
    "age": 22,
    "course": "Data Science"
}

print(student)
```

### Access Values

```python
print(student["name"])
print(student.get("age"))
```

### Dictionary Methods

```python
student["age"] = 23         # Update
student["city"] = "Lahore"   # Add
student.pop("course")       # Remove

print(student)
```

### Looping Dictionary

```python
for key, value in student.items():
    print(key, ":", value)
```
