### Shorthand if ... else

```python
a = 5
b = 2
if a > b : print("a is greater than b ")
```

### shorthand if ...... else

```python
a = 300
b = 20
print("A") if a > b else print("B") 
```

### Assign a value with if ...... else

```python
# Assign a value with if ... else
a = 10
b = 20
bigger = a if a > b else b
print(f"bigger is {bigger}")
```
