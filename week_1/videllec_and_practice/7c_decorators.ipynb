### Multiple Decorator Calls

```python
def changecase(func):
    def innerfun():
        return func().upper()
    return innerfun

@changecase
def myfunc():
    return "hello world"

@changecase
def mysecondfun():
    return "kesa ho"

print(myfunction())
print(myfunction()) 
```

### Arguments in the Decorated Function

```python
def changecase(func):
  def myinner(x):
    return func(x).upper()
  return myinner

@changecase
def myfunction(nam):
  return "Hello " + nam

print(myfunction("John"))
```

```python

def greeting(fun):
    def modifiedfun():
        print("greeting")
        fun()
        print("goodbye")
    return modifiedfun


@greeting
def func():
    print("Hello world")
    
func()
```

### *args and **kwargs in Decorators

```python
def addationalfunctiondecorartor(func):
    def inner_functionality(*args,**kwargs):
        print(f"*args are : {args}")
        print(f"**kwargs are : {kwargs}")
        result =  func(*args,**kwargs)
        return result.upper()
    return inner_functionality
    


@addationalfunctiondecorartor
def mydetail(name, age, city, course):
    return f"Hello {name}, you are {age} years old, you live in {city}, and you study {course}."

mydetail("talha",23,city="lahore",course="django")
```

### Decorator With Arguments
Decorators can accept their own arguments by adding another wrapper level.

```python
def changecase(n):
  def changecase(func):
    def myinner():
      if n == 1:
        a = func().lower()
      else:
        a = func().upper()
      return a
    return myinner
  return changecase

@changecase(1)
def myfunction():
  return "Hello Linus"

print(myfunction())
```
