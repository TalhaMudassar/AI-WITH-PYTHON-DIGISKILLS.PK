```python
import matplotlib.pyplot as plt

# Sample data

days = [1,2,3,4,5,6,7]
sales = [120,150,180,170,200,220,210]

plt.plot(days,sales,
         marker='o',
         linewidth=2,
         alpha=0.8,
         label=("Daily Sales"))

plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Sales Trend Over a Week")
plt.legend()
plt.show()
```

### Scatter plot

```python
import matplotlib.pyplot as plt

# Sample data
temperature = [20, 22, 25, 27, 30, 32, 35]
ice_cream_sales = [50, 65, 80, 90, 120, 140, 160]

plt.scatter(temperature, ice_cream_sales,
            alpha=0.7,
            marker='x',
            label="Sales vs Temp")

plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales")
plt.title("Relationship Between Temp & Sales")
plt.legend()
plt.show()
```
