## Arthematic Operation(+, -, *, /, //, %, **)

```python
number = 10

if number % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
number = 2
result = number ** 3
print(f"Result are :{result}")
```

```python
number = 10
result = number / 3
print("Results are:",result)
```

```python
number = 10
result = number // 3
print("Results are:",result)
```

## Comparison Operators (==, !=, >, <, >=, <=)

```python
number = 10
if number % 2 == 0:
    print("number are even")
else:
    print("number are odd")
```

## Logical Operators (and, or, not)

```python
age = 20
has_cnic = True

if age >=18 and has_cnic:
    print("You are eligible to open a bank account")
else:
    print("You are not eligible to open a bank account")
```

```python
age = 20
has_cnic = True

if age <=18 or has_cnic:
    print("You are eligible to open a bank account")
else:
    print("You are not eligible to open a bank account")
```

```python
is_banned = False

# The 'not' operator reverses the False to True
# So, it reads as: "if the user is NOT banned"
if not is_banned:
    print("Access granted. Welcome to the dashboard!")
else:
    print("Access denied. Your account has been banned.")
```

## Membership Operators (in, not in)

```python
name = "Talha"
if "a" in name:
    print(f"{name} Contain vowel letter")
```

```python
allowed_users = ["admin", "moderator", "superuser"]
current_user = "guest"

# The condition checks if "guest" is missing from the list
if current_user not in allowed_users:
    print("Access denied. You do not have the required permissions.")
else:
    print("Welcome to the system!")
```

## Identity Operators (is, is not)

```python
active_user = None

# Using 'is' to check the exact identity of the object
if active_user is None:
    print("No user is currently logged in.")
else:
    print("Welcome back, user!")
```
