```python
class Car:
    color = "red"
    
    def drive(self):
        print(f"{self.color} car is moving ! ")
        
    def setColor(self,new_color):
        self.color = new_color

c1 = Car()

c1.setColor("green")

c1.drive()        
```

## __init__() constructor function

```python
class Dog:
    def __init__(self):
        self.name = "buddy"
        
    def bark(self):
        print(f"{self.name} says Woof! ")
```

```python
dog1 = Dog()
dog1.bark()
```

```python
class Car:
    def __init__(self,brand,color):
        self.brand = brand
        self.color = color
        
car1 = Car("Toyota","Red")
print(car1.brand)
print(car1.color)
```

```python
car1.color = "Blue"
car2 = Car("Honda","White")
```

```python
print(car2.brand)
print(car2.color)
```

# Instance Variables and Instance Methods

```python
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def display_info(self):
        print(f"Student Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Grade: {self.grade}")

    def is_eligible(self):
        if self.age >= 15:
            print(self.name, "is eligible for admission.")
        else:
            print(self.name, "is not eligible for admission.")
```

### Accessing Object Attributes & Methods

```python
# Creating Objects
student1 =  Student("Ali",16,"10th")
student2 = Student("Sara",14,"18th")
```

```python
# Accessing Attributes
print(student1.name)
print(student1.grade)
```

```python
# Calling Methods
student1.display_info()
student1.is_eligible()
```

```python
student2.display_info()
student2.is_eligible()
```
