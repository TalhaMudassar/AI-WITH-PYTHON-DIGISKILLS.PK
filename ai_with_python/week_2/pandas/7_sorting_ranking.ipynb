### Sorting anD Ranking in pandas

```python
import pandas as pd
```

```python
df = pd.read_csv(
    "retail_sales.csv",
    parse_dates=["Date"],
    dtype={
        "Category":"category",
        "Region":"category"
    }
)

df.head()
```

```python
# Sort by Age ascending, than Salary desending 
df_sorting = df.sort_values(by=['Quantity','Sales'],ascending=[False,True])
print("Sorted the dataframe")
df_sorting.head()
```

```python
#Rank salary within the Age Groups
df['Sales_rank'] = df.groupby('Quantity')['Sales'].rank(method='dense',ascending=False)
print("\n Sales rank with in Quantity Group :\n",df)
```
