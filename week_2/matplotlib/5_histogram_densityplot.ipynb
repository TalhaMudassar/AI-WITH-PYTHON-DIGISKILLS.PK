```python
import matplotlib.pyplot as plt
import numpy as np
group_a = np.random.normal(70,10,1000)
group_b = np.random.normal(80,12,1000)
plt.hist(group_a, bins=30, density=True,alpha=0.6,label="GroupA")
plt.hist(group_b, bins=30, density=True,alpha=0.6,label="GroupB")

plt.xlabel("Score")
plt.ylabel("Density")
plt.title("Distribution Comparsion")
plt.legend()
plt.show()
```
