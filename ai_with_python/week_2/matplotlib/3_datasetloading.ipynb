```python
# Load Dataset
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    "ecommerce_sales_analytics_5000.csv",
    parse_dates=["order_date"]
)
df = df.sample(n=100)
```

```python
# Line Plot - Monthly Revenue Trend
monthly_revenue = (
    df
    .set_index("order_date")["revenue"]
    .resample("ME")
    .sum()
)
```

```python
# Line Plot - Monthly Revenue Trend
monthly_revenue = (
    df
    .set_index("order_date")["revenue"]
    .resample("ME")
    .sum()
)

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(monthly_revenue.index, monthly_revenue.values, linewidth=2)

ax.set_title("Monthly Revenue Trend")
ax.set_xlabel("Month")
ax.set_ylabel("Total Revenue")

# Explanation: Line connects revenue over time and Shows growth or decline clearly
```

```python
# Scatter Plot - Discount vs Revenue
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["unit_price"], df["revenue"], alpha=0.5)

ax.set_title("Unit Price vs Revenue")
ax.set_xlabel("Unit Price")
ax.set_ylabel("Revenue")

# Each point = one order and Transparency (alpha) helps with overlapping points
```
