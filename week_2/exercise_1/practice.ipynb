### Apply loops and conditional statements to solve simple problems.

```python
total_sales = 0.0
num_employees = 5

for i in range(1, num_employees+1):
    sales = float(input(f"Enter daily sales for employee {i} (Rs.):"))
    
    if sales < 50000:
        tax = sales * 0.05
    elif 50000 <= sales <= 100000:
        tax = sales * 0.10
    else:
        tax = sales * 0.15
        
    total_sales += sales
    
    print(f"Sales Amount: Rs. {sales:.2f}")
    print(f"Tax Amount: Rs. {tax:.2f}\n")
    
average_sales = total_sales / num_employees


print("--- Summary ---")
print(f"Total Sales: Rs. {total_sales:.2f}")
print(f"Average Sales: Rs. {average_sales:.2f}")
```

### Create and manipulate NumPy arrays.

```python
import numpy as np 

# 1. Create a NumPy array containing the marks
marks_array = np.array([10,20,30,40,50,60])

# 2. Display the Results 
print("Original Marks Array:", marks_array)
print("Data Type:", marks_array.dtype)
print("-" * 30)


# 3. Convert the array into float data type
float_array = marks_array.astype(float)


# 4. Display the converted array
print("Converted Float Array:",float_array)
print("New Data Type:",float_array.dtype)
```

```python
class Customer:
    def __init__(self, name):
        self.name = name

class Account:
    def __init__(self, customer, initial_balance=0.0):
        self.customer = customer
        self.balance = initial_balance
        print(f"Account successfully created for {self.customer.name}!")

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited Rs. {amount}. New balance is Rs. {self.balance}.")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew Rs. {amount}. Remaining balance is Rs. {self.balance}.")
        elif amount > self.balance:
            print("Withdrawal failed: Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")

    def display_balance(self):
        print(f"Current Balance for {self.customer.name}: Rs. {self.balance}")


# --- System Functionality Demonstration ---

# 1. Create a customer account
customer_info = Customer("Talha Mudassar")
my_account = Account(customer_info, initial_balance=5000.0)

# 2. Display the current balance
my_account.display_balance()

# 3. Deposit money
my_account.deposit(2500.0)

# 4. Withdraw money
my_account.withdraw(1200.0)

# Display final balance
my_account.display_balance()
```
