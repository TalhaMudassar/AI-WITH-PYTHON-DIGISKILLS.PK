```python
import pandas as pd

df = pd.read_csv(
    "retail_sales.csv",
    parse_dates=["Date"],        # convert Date column to datetime
    dtype={
        "Category": "category"
    }
)
df.info()

```

```python
# Convert Age to numeric
df['Profit'] = pd.to_numeric(df['Profit'])
df.info()
```

```python

# Convert City to category
df['Region'] = df['Region'].astype('category')
df.info()
print(df['Region'].memory_usage(deep=True))
```
