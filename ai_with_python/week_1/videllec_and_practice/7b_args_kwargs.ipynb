# *args and **kwargs
### By default, a function must be called with the correct number of arguments.
### However, sometimes you may not know how many arguments that will be passed into your function.
### *args and **kwargs allow functions to accept a unknown number of arguments.

## *args?

```python
def my_function(*args):
  print("Type:", type(args))
  print("First argument:", args[0])
  print("Second argument:", args[1])
  print("All arguments:", args)
  
  
my_function("first","second","third","fourth")
    
```

```python
# Using *args with Regular Arguments
def my_function(name,*args):
    print(f"Name are {name}")
    print("skills are",args)
    
my_function("talha","python","js","html","css")
```

# Kwargs

```python
def my_function(**kwargs):
    print("responses are ",kwargs)
    
    
my_function(name="talha",age=24,score=45,grade="A")
```

```python
# # Using **kwargs with Regular Arguments
# You can combine regular parameters with **kwargs.

# Regular parameters must come before **kwargs:
def kwarg_regular(name,**kwargs):
    print("name are ",name)
    print("kwargs are ",kwargs)
    
kwarg_regular("talha",age=23,score=89,grade="A")
    
```

## Combining *args and **kwargs
### You can use both *args and **kwargs in the same function.

The order must be:

1. regular parameters
2. *args
3. **kwargs

```python
def mixing_function(regularparm,*args,**kwargs):
    print("regualar arguments are :",regularparm)
    print("*arguments are :",args)
    print("**kwargs are :",kwargs)
mixing_function("am regular","first","second","third",final="finalkwargs")
```

# Unpacking Arguments
### The * and ** operators can also be used when calling functions to unpack (expand) a list or dictionary into separate arguments.

---
### Unpacking Lists with *

```python
def my_Fun(a,b,c):
    print(a,b,c)
    
mylist = [1,2,3]

my_Fun(*mylist)
```

---
### Unpacking Dictionaries with **

```python
def myfunction(fname,lname):
    print(fname,lname)
    
    
my_dict = {"fname":"talha","lname":"mudassar"}
myfunction(**my_dict)
```
