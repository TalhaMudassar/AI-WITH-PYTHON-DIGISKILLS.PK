```python
import numpy as np 
```

```python
# Simulate 100k coins  flips
rng = np.random.default_rng()
flips = rng.integers(0,2,size=100000) # 0= tails ,   1 = heads
prob_head = np.mean(flips)
print("Establised probablities of head", prob_head)
```

```python
# if we want fix no of heads come so we use seed 
rng = np.random.default_rng(seed=43) # here we use the seeds 
flips = rng.integers(0,2,size=100000) # 0= tails ,   1 = heads
prob_head = np.mean(flips)
print("Establised probablities of head", prob_head)
```
