# Tuples METHODS 

### .count(x)
Counts exactly how many times a specific value appears in the tuple.

```python
tuple_1 = (1,2,3,4,5,6,2,3,4,5,2,2,2,6)

print(tuple_1.count(2))
```

```python
# A tuple recording the results of coin flips
coin_flips = ("Heads", "Tails", "Heads", "Heads", "Tails")
# Count how many times "Heads" appears
heads_count = coin_flips.count("Heads")
print("You flipped Heads", heads_count, "times.")
# Output: You flipped Heads 3 times.
```

### .index(x)
Searches for a specific value and returns the index number of its very first occurrence.

```python
tuple_1 = (1,2,3,10,4,5,6,7,8)

store = tuple_1.index(10)

print(store)
```
