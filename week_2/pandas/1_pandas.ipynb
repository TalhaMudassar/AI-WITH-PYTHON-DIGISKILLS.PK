### import pandas

```python
import pandas as pd
```

### Create a series

```python
s = pd.Series([10,20,20,30,40], index=['a','b','c','d','e'])
print(s)
```

### Index vs Values

```python
print("Values",s.values)
print("Indexes",s.index)
print("Dtype",s.dtype)
```

```python
print(s.tail)
```
